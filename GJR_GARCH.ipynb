```python
import numpy as np
import pandas as pd
from arch import arch_model


def gjr_model(data, p=1, o=1, q=1, dist="t"):

    # Clean missing values
    returns = data.dropna()

    # Model specification
    model = arch_model(
        returns, mean="Constant", vol="GARCH", p=p, o=o, q=q, dist=dist
    )

    # Model fitting
    result = model.fit(disp="off")


    # 1-day-ahead forecast
    forecast = result.forecast(horizon=1)
    variance_1d = forecast.variance.iloc[-1, 0]
    vol_daily = np.sqrt(variance_1d)
    vol_annual = vol_daily * np.sqrt(252)

    print(f"Daily_Vol : {vol_daily :.4f}%")
    print(f"Ann_Vol  : {vol_annual :.4f}%")

    return result
```
